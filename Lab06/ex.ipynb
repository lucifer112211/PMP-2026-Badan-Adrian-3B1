```python
import scipy.stats as stats
import arviz as az
import matplotlib.pyplot as plt

n = 10
total_apeluri = 180

alpha = 4
beta = 0.2

alpha_post = alpha + total_apeluri
beta_post = beta + n

samples = stats.gamma.rvs(
    a=alpha_post,
    scale=1 / beta_post,
    size=100000
)

hdi = az.hdi(samples, prob=0.94)

print("HDI 94%:", hdi)
```
