```python
import numpy as np

n = 100000

pozitiv = 0
bolnav_pozitiv=0

for i in range(n):
    bolnav = np.random.random() <0.01

    if bolnav:
        test_pozitiv = np.random.random() <0.95
    else:
        test_pozitiv = np.random.random() <0.1
    if test_pozitiv:
        pozitiv+=1
        if bolnav:
            bolnav_pozitiv+=1

prob = bolnav_pozitiv/pozitiv

print(prob)
    


```
