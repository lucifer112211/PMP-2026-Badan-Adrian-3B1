Citirea studentilor din fisierul .csv:

```python
import csv
import random

fisier = "studenti.csv"
k = 5

studenti = []

with open(fisier,"r",encoding="utf-8") as f:
    reader = csv.reader(f)

    for line in reader:
        studenti.append(line[0])
```

Alegerea studentilor:

```python
esantion = random.sample(studenti,k)

print("Au fost alesi studentii: ")

for student in esantion:
    print(student)
```

Nota: ar putea fi adaugata o verificare intre k si numarul de studenti.
