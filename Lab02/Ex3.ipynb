Calcularea probabilitatilor si a timpului de servire:

```python
import numpy as np
import matplotlib.pyplot as plt

lambdas = np.array([3,6,4])
probabilitati = lambdas/sum(lambdas)

n = 10000
X = []

for i in range(n):
    frizer = np.random.choice([0,1,2],p=probabilitati)
    timp = np.random.exponential(scale = 1/lambdas[frizer])
    X.append(timp)

print("Media (h) :", np.mean(X))
print("stdev: ",np.std(X))

plt.hist(X, bins=50, density=True, edgecolor="black")

plt.xlabel("Timpul de servire X (ore)")
plt.ylabel("Densitate")
plt.title("Densitatea lui X")

plt.show()
```
