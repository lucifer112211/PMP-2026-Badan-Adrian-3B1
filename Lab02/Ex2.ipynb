Functia pentru simularea unui joc:

```python
import numpy as np
import matplotlib.pyplot as plt


def simuleaza_joc(p=0.5):
    N = 0
    S = 0

    while True:
        N = N + 1
        moneda = np.random.random() #numar intre 0 si 1

        if moneda < p:
            zar = np.random.randint(1,7)
            S = S + (zar-3)
            break
        else:
            S = S - 0.5
    
    return N,S
```

Simularea unui singur joc:

```python
N,S = simuleaza_joc()

print("Pasi ", N)
print("suma ",S)
```

Simularea unui numar mare de jocuri:

```python
simulari = 100000
sume = []

for i in range(simulari):
    N,S = simuleaza_joc()
    sume.append(S)

print("Avg(S): ",np.mean(sume))

plt.hist(sume,bins=30,edgecolor="black")
plt.xlabel("S")
plt.ylabel("frecv")
plt.title("Distributia lui S, p=1/2")
plt.show()
```

Moneda "masluita":

```python


for p in [0.3,0.7]:
    
    sume = []
    for i in range(simulari):
        N,S = simuleaza_joc(p)
        sume.append(S)

    print("Avg(S): ",np.mean(sume))

    plt.hist(sume,bins=30,edgecolor="black")
    plt.xlabel("S")
    plt.ylabel("frecv")
    plt.title("Distributia lui S, p: "+str(p))
    plt.show()

```
