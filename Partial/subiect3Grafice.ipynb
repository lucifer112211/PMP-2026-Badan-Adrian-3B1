```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import beta

alpha, b = 2, 8

x = np.linspace(0, 1, 200)
y = beta.pdf(x, alpha, b)

plt.plot(x, y)
plt.title(f'Beta({alpha}, {b})')
plt.show()
```
