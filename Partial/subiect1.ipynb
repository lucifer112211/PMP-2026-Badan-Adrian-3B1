Robotul la inceput are o pozitie uniform distribuita, deci poate fi in oricare dintre camerele 1,2,3,4

```python
import numpy as np
from hmmlearn.hmm import CategoricalHMM

#avem 4 stari, iar cele 3 culori ne dau 3 features
model = CategoricalHMM(
    n_components=4,
    n_features=3,
    init_params=""
)

model.startprob_ = np.array([
    0.25,0.25,0.25,0.25
])
#tranzitia de la o camera la alta
#exemplu : camera 1 -> 2 si camera 4 cu prob 0.4, ramane pe loc cu prob 0.2
model.transmat_ = np.array([
    [0.2,0.4,0.0,0.4], #1->1,2,4
    [0.4,0.2,0.4,0.0],#2->1,3,2
    [0.0,0.4,0.2,0.4], #3 -> 2,3,4
    [0.4,0,0.4,0.2]#4-> 3,4,1
])
#matricea de emisie, data in enunt, pe rand e camera , pe coloana e observatia
model.emissionprob_ = np.array([
    [0.7,0.2,0.1],
    [0.2,0.6,0.2],
    [0.1,0.3,0.6],
    [0.5,0.2,0.3] 
])
```

Vom aplica modelul viterbi pentru secventa: red,green,blue,blue,red

```python
cod = {"Red": 0,"Green": 1,"Blue": 2}

observatii = ["Red","Green","Blue","Blue","Red"]

X = np.array([[cod[x]] for x in observatii])

log_prob, secventa = model.decode(X,algorithm="viterbi")

nume_stari = {0: "Camera 1",1: "Camera 2",2: "Camera 3",3:"Camera 4"}

secventa_nume = [nume_stari[s] for s in secventa]

print("Secventa:")
print(secventa_nume)

print("Log-probabilitate:", log_prob)
print("Probabilitate:", np.exp(log_prob))
```

Repetam experimentul creand alt model: de data asta, probabilitatile de tranzitie sunt diferite
