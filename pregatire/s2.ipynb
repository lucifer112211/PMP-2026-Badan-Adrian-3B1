```python
import numpy as np
from hmmlearn.hmm import CategoricalHMM

model = CategoricalHMM(
    n_components=3,
    n_features=3,
    init_params=""
)

model.startprob_ = np.array([
    0.4,0.3,0.3
])

model.transmat_ = np.array([
    [0.6,0.3,0.1],
    [0.2,0.7,0.1],
    [0.3,0.2,0.5]
])

model.emissionprob_ = np.array([
    [0.1,0.7,0.2],
    [0.05,0.25,0.7],
    [0.8,0.15,0.05] 
])
```

b: ### b) Forward

Pentru secventa de observatii:

$$
[M,H,L]
$$

folosim recurenta Forward:

$$
\alpha_1(j)=\pi_jb_j(o_1)
$$

si:

$$
\alpha_t(j)
=
b_j(o_t)\sum_i\alpha_{t-1}(i)a_{ij}
$$

Probabilitatea finala a secventei este:

$$
P(M,H,L)
=
\sum_j\alpha_3(j)
=
0.02111725
$$

Deci probabilitatea observarii secventei este de aproximativ $2.11\%$.

```python
pi = np.array([0.4, 0.3, 0.3])

A = np.array([
    [0.6, 0.3, 0.1],
    [0.2, 0.7, 0.1],
    [0.3, 0.2, 0.5]
])

B = np.array([
    [0.1, 0.7, 0.2],   # W
    [0.05, 0.25, 0.7], # R
    [0.8, 0.15, 0.05]  # S
])
```

```python
obs = [1, 2, 0]  # M, H, L

alpha = pi * B[:, obs[0]]

print("alpha 1 =", alpha)

for t in range(1, len(obs)):
    alpha = B[:, obs[t]] * (alpha @ A)
    print(f"alpha {t + 1} =", alpha)

probabilitate = np.sum(alpha)

print("P(M,H,L) =", probabilitate)
```

```python
cod = {"Medium": 1,"High": 2,"Low": 0}

observatii = ["Medium","High","Low"]

X = np.array([[cod[x]] for x in observatii])

log_prob, secventa = model.decode(X,algorithm="viterbi")

nume_stari = {0: "W",1: "R",2: "S"}

secventa_nume = [nume_stari[s] for s in secventa]

print("Secventa:")
print(secventa_nume)

print("Log-probabilitate:", log_prob)
print("Probabilitate:", np.exp(log_prob))
```

```python
rng = np.random.default_rng(42)

nr_simulari = 10000
aparitii = 0

secventa_cautata = [1, 2, 0]  # M, H, L

for _ in range(nr_simulari):

    # starea initiala
    stare = rng.choice(3, p=pi)

    observatii_generate = []

    # prima observatie
    observatie = rng.choice(3, p=B[stare])
    observatii_generate.append(observatie)

    # urmatoarele doua momente
    for t in range(1, 3):
        stare = rng.choice(3, p=A[stare])

        observatie = rng.choice(
            3,
            p=B[stare]
        )

        observatii_generate.append(observatie)

    if observatii_generate == secventa_cautata:
        aparitii += 1

prob_empirica = aparitii / nr_simulari

print("Aparitii:", aparitii)
print("Probabilitate empirica:", prob_empirica)
print("Probabilitate Forward:", probabilitate)
```
