Reteaua bayesiana se scoate din probabilitati:
O-> H, O->W, H->E W->R H->R R->C

```python
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD

model = DiscreteBayesianNetwork([("O","H"),("O","W"),("H","E"),("W","R"),("H","R"),("R","C")])

cpd_o = TabularCPD(variable ="O",variable_card = 2, values=[[3/10],[7/10]])
print (cpd_o)

cpd_h = TabularCPD(variable = "H", variable_card=2,
    values = [
        [0.9,0.2],
        [0.1,0.8]
    ],

    evidence = ["O"],
    evidence_card=[2]
)
print(cpd_h)
cpd_w = TabularCPD(variable = "W", variable_card=2,
    values = [
        [0.1,0.6],
        [0.9,0.4]
    ],

    evidence = ["O"],
    evidence_card=[2]
)
print(cpd_w)

cpd_r = TabularCPD(variable = "R", variable_card=2,
    values = [
        [0.6,0.9,0.3,0.5],
        [0.4,0.1,0.7,0.5]
    ],

    evidence = ["H","W"],
    evidence_card=[2,2]
)
print(cpd_r)

cpd_e = TabularCPD(variable = "E", variable_card=2,
    values = [
        [0.8,0.2],
        [0.2,0.8]
    ],

    evidence = ["H"],
    evidence_card=[2]
)
print(cpd_e)


cpd_c = TabularCPD(variable = "C", variable_card=2,
    values = [
        [0.85,0.4],
        [0.15,0.6]
    ],

    evidence = ["R"],
    evidence_card=[2]
)
print(cpd_c)

model.add_cpds(cpd_o,cpd_h,cpd_w,cpd_r,cpd_e,cpd_c)
print(model.check_model())
```

pentru b calculam folosind reteaua

```python
from pgmpy.inference import VariableElimination

infer = VariableElimination(model)
posterior_H = infer.query(variables=["H"],evidence={"C":0})
print(posterior_H)

posterior_E = infer.query(variables=["E"],evidence={"C":0})
print(posterior_E)

posterior_HW = infer.query(variables=["H","W"],evidence={"C":0})
print(posterior_HW)
```
